# **Building a Digits Classification Model**

## **Import Libraries**

```python
import matplotlib.pyplot as plt
from sklearn import datasets
```

## **Loading the dataset**

```python
digits = datasets.load_digits()
digits
```

```python
## Images data
digits.images.shape
```

## **Visualiztion of digits**

```python
_, axes = plt.subplots(nrows=1, ncols=10, figsize=(16,4))
for ax, image, label in zip(axes, digits.images, digits.target):
  ax.set_axis_off()
  ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
  ax.set_title('Training: %i' % label)

```

## **Data Processing**

```python
# Examining the image data
digits.images
```

```python
#Examining the labels
digits.target
```

```python
# flatten the images
import pandas as pd
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

#Displaying the contents of the processed x data matris as a DataFrame
pd.DataFrame(data)
```

```python
# Dimensions of original image data
digits.images.shape
```

```python
# Dimensions of reshaped data
data.shape
```

As we can see, the processed digits.images data is the same as the provided digits.data

```python
pd.DataFrame(digits.data)
```

## **Data Splitting**

```python
# split data into 50% train and test subsets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split( data, digits.target, test_size=0.5, shuffle =False)
```

## **Model Building**

```python
# Support vector classifier
from sklearn.svm import SVC

svm = SVC(gamma=0.001, random_state=42)
# Learn the digits on the train subset
svm.fit(X_train, y_train)
```

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=500, random_state=42)
rf.fit(X_train, y_train)
```

## **Applying model to make Predictions**

```python
# Predict the value of the digit on the test subset
y_test_pred_svm = svm.predict(X_test)
y_test_pred_rf = rf.predict(X_test)
```

## **Visualization of the predicted digits**

```python
#SVM
_, axes = plt.subplots(nrows=1, ncols=10, figsize=(16,4))
for ax, image, actual, prediction in zip(axes, X_test, y_test, y_test_pred_svm):
  ax.set_axis_off()
  image = image.reshape(8,8)
  ax.imshow(image,cmap=plt.cm.gray_r, interpolation='nearest')
  ax.set_title(f'Prediction: {prediction}\n Actual: {actual}')
```

```python
#RF
_, axes = plt.subplots(nrows=1, ncols=10, figsize=(16,4))
for ax, image, actual, prediction in zip(axes, X_test, y_test, y_test_pred_rf):
  ax.set_axis_off()
  image = image.reshape(8,8)
  ax.imshow(image,cmap=plt.cm.gray_r, interpolation='nearest')
  ax.set_title(f'Prediction: {prediction}\n Actual: {actual}')
```

## **Model**

```python
from sklearn.metrics import accuracy_score
accuarcy_svm = accuracy_score(y_test, y_test_pred_svm)
accuracy_rf = accuracy_score(y_test, y_test_pred_rf)

print('Accuracy (SVM):', round(accuarcy_svm, 3) )
print('Accuracy (RF): ', round(accuracy_rf, 3))
```

## **Mathews correlation coefficient**

```python
from sklearn.metrics import matthews_corrcoef

mcc_svm = matthews_corrcoef(y_test, y_test_pred_svm)
mcc_rf = matthews_corrcoef(y_test, y_test_pred_rf)

print('MCC (SVM): ', round(mcc_svm, 3))
print('MCC (RF): ', round(mcc_rf, 3))
```

### **Confusion Matrix**

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
# Support vector machine
disp = ConfusionMatrixDisplay.from_estimator(svm, X_test, y_test)
disp.figure_.suptitle('Confusion Matrix')
print(f'Confusion matrix: \n {disp.confusion_matrix}')
plt.show()
```

```python
# Random Forest
disp = ConfusionMatrixDisplay.from_estimator(rf, X_test, y_test)
disp.figure_.suptitle('Confusion Matrix')
print(f'Confusion Matrix:/n {disp.confusion_matrix}')
plt.show()
```

```python
# Random forest
disp = ConfusionMatrixDisplay.from_estimator(rf, X_test, y_test, cmap='cool')
disp.figure_.suptitle('Confusion Matrix')
print(f'Confusion Matrix:/n {disp.confusion_matrix}')
plt.show()
```
